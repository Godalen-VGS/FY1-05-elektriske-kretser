Enkle kretser kan du regne ut med Ohms lov og reglene for serie- og parallellkobling. Men har kretsen flere batterier, eller motstander som verken står i serie eller parallell, må du bruke **Kirchhoffs lover**. De gir et **likningssett**, og der er sympy et perfekt verktøy.

## Regler

| Regel | Formel |
| --- | --- |
| Ohms lov | $U = R\,I$ |
| Seriekobling | $R = R_1 + R_2 + \dots$ |
| Parallellkobling | $\dfrac{1}{R} = \dfrac{1}{R_1} + \dfrac{1}{R_2} + \dots$ |
| Effekt | $P = U\,I = R\,I^2$ |
| Kirchhoffs 1. lov (strøm) | Strømmen inn i et knutepunkt = strømmen ut |
| Kirchhoffs 2. lov (spenning) | Summen av spenningene rundt en lukket sløyfe er 0 |

## Eksempel: Kirchhoff med sympy

Et batteri på 9,0 V er koblet til $R_1 = 100\,\Omega$ i serie med $R_2 = 200\,\Omega$ og $R_3 = 300\,\Omega$ i parallell.

Vi kaller strømmene $I_1$ (gjennom $R_1$), $I_2$ og $I_3$, og setter opp:

- Knutepunkt: $I_1 = I_2 + I_3$
- Sløyfe gjennom batteri, $R_1$ og $R_2$: $9{,}0 - 100\,I_1 - 200\,I_2 = 0$
- Sløyfe gjennom $R_2$ og $R_3$: $200\,I_2 - 300\,I_3 = 0$

```python
import sympy as sp

I1, I2, I3 = sp.symbols('I1 I2 I3')

likninger = [
    sp.Eq(I1, I2 + I3),
    sp.Eq(9.0 - 100*I1 - 200*I2, 0),
    sp.Eq(200*I2 - 300*I3, 0),
]
print(sp.solve(likninger, [I1, I2, I3]))
```

```python
# Kjør denne cellen for å sjekke svarene dine. Du trenger ikke forstå koden.
def naer(a, b, rel=1e-3):
    """Sant hvis tallet a er innenfor rel (relativt) av b."""
    try:
        a, b = float(a), float(b)
    except (TypeError, ValueError):
        return False
    if b == 0:
        return abs(a) < 1e-9
    return abs(a - b) <= rel * abs(b)


def S(tekst):
    """Gjør tekst om til et sympy-uttrykk, f.eks. S("x**2 - 1")."""
    import sympy
    return sympy.sympify(tekst, locals={n: sympy.Symbol(n) for n in ("N", "O", "Q", "S")})


def lik(a, b):
    """Sant hvis a og b er matematisk like.

    Lister og mengder sammenlignes uten hensyn til rekkefølge, tupler (koordinater) i rekkefølge.
    """
    import sympy
    if a is None:
        return False
    if isinstance(b, tuple):
        if not isinstance(a, (tuple, list, sympy.Tuple)) or len(a) != len(b):
            return False
        return all(lik(x, y) for x, y in zip(a, b))
    if isinstance(b, (list, set)):
        if isinstance(a, sympy.FiniteSet):
            a = list(a)
        if not isinstance(a, (list, tuple, set)) or len(a) != len(b):
            return False
        rest = list(a)
        for y in b:
            for i, x in enumerate(rest):
                if lik(x, y):
                    del rest[i]
                    break
            else:
                return False
        return True
    if isinstance(b, dict):
        if not isinstance(a, dict) or len(a) != len(b):
            return False
        a = {str(k): v for k, v in a.items()}
        return all(str(k) in a and lik(a[str(k)], v) for k, v in b.items())
    try:
        a, b = sympy.sympify(a), sympy.sympify(b)
    except (sympy.SympifyError, TypeError):
        return False
    vanlige = lambda u: u.subs({s: sympy.Symbol(s.name) for s in u.free_symbols})
    a, b = vanlige(a), vanlige(b)
    if isinstance(a, sympy.Set) or isinstance(b, sympy.Set):
        return a == b
    if getattr(b, "is_Relational", False):
        return a == b
    try:
        return sympy.simplify(a - b) == 0
    except TypeError:
        return a == b

SJEKKER = [
    ('1_serie', 'serie([2, 3, 5]) skal gi 10', 'naer(serie([2, 3, 5]), 10)'),
    ('1_parallell', 'parallell([6, 12]) skal gi 4', 'naer(parallell([6, 12]), 4) and naer(parallell([10, 10, 10]), 10/3)'),
    ('2', 'I_tot, I_2 og I_3 skal være 1.5 A, 1.0 A og 0.5 A', 'naer(I_tot, 1.5) and naer(I_2, 1) and naer(I_3, 0.5)'),
    ('3', 'strommer skal være I1 = 21/11 A, I2 = -6/11 A og I3 = 15/11 A', "lik(strommer, {'I1': S('21/11'), 'I2': S('-6/11'), 'I3': S('15/11')})"),
    ('4', 'effekt_motstand(4, 1.5) skal gi 9 W', 'naer(effekt_motstand(4, 1.5), 9) and naer(effekt_motstand(100, 0.1), 1)'),
    ('5_lik', 'belastet_deler(12, 1000, 1000, 1000) skal gi 4 V', 'naer(belastet_deler(12, 1000, 1000, 1000), 4)'),
    ('5_stor_last', 'belastet_deler med svært stor last skal nærme seg 6 V', 'naer(belastet_deler(12, 1000, 1000, 1e12), 6, rel=1e-6)'),
]


def test_alt():
    feil = 0
    for _, beskrivelse, uttrykk in SJEKKER:
        try:
            ok = eval(uttrykk)
        except Exception as e:
            ok = False
            beskrivelse += f" (krasjet: {type(e).__name__}: {e})"
        if not ok:
            print("❌", beskrivelse)
            feil += 1
    if feil == 0:
        print(f"✅ Alle testene bestått! ({len(SJEKKER)} av {len(SJEKKER)})")
    else:
        print(f"\n{len(SJEKKER) - feil} av {len(SJEKKER)} tester bestått.")


test_alt()
```
